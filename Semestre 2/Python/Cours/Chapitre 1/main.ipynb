Ecrire un programe qui demande a l'utilisateur de saisir 3 note et 3 coef respectue. Determine et affiche  la moyenne ponderer

```python
note1= float(input("Donner la note 1 : "))
note2= float(input("Donner la note 2 : "))
note3= float(input("Donner la note 3 : "))
coef1= int(input("Donner la coefficient 1 : "))
coef2= int(input("Donner la coefficient 2 : "))
coef3= int(input("Donner la coefficient 3 : "))
somme_coef= coef1+coef2+coef3
somme_note= note1+note2+note3
moyenne= somme_note/somme_coef
print(f"La moyenne de la note : {moyenne: .2f}/20")
```
